```python
import json
import math

with open('data.json', 'r') as file:
    data = json.load(file)
```

El producto con más ingresos

```python
ingreso_producto={}
for item in data["ventas"]:
    producto = item["producto"]
    ingreso = item["cantidad"] * item["precio_unitario"]
    if producto not in ingreso_producto:
        ingreso_producto[producto]=0
    ingreso_producto[producto] += ingreso

prod_mas_ingresos= ""
max_ingresos=-math.inf
for prod, ingreso in ingreso_producto.items():
    if ingreso>max_ingresos:
        max_ingresos=ingreso
        prod_mas_ingresos=prod
print("-El producto con mas ingresos es:",  prod_mas_ingresos, ", con" , max_ingresos, " €.")
```

Filtra las ventas por un vendedor concreto

```python
vendedor_buscado = "Ana"
ventas_filtradas = []

print("-Ventas del vendedor", vendedor_buscado + ":")
for item in data["ventas"]:
    if item["vendedor"] == vendedor_buscado:
        ventas_filtradas.append(item)
        print("  ID:", item["id"], "- Producto:", item["producto"], "- Cantidad:", item["cantidad"], "- Precio:", item["precio_unitario"], "€")
```

Calcula el ticket medio

```python
total_ingresos = 0
for item in data["ventas"]:
    ingreso = item["cantidad"] * item["precio_unitario"]
    total_ingresos += ingreso

num_ventas = len(data["ventas"])
ticket_medio = total_ingresos / num_ventas
print("-Ticket medio:", round(ticket_medio, 2), "€")
```
